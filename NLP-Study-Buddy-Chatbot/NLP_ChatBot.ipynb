```python
import nltk
import random
import json
import re

from nltk.chat.util import reflections
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# ===========================
# DOWNLOAD NLTK RESOURCES
# ===========================

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab') # Download 'punkt_tab' for specific tokenization needs

# ===========================
# KNOWLEDGE BASE
# ===========================

knowledge_base = {

    "greeting": {
        "patterns": ["hello", "hi", "hey", "good morning", "good evening"],
        "responses": [
            "Hello! I'm Enisteen. How can I help you?",
            "Hi there! Ready to learn AI and NLP?",
            "Hey! Ask me anything about AI, ML, DL or NLP."
        ]
    },

    "ai": {
        "patterns": ["ai", "artificial intelligence", "what is ai"],
        "responses": [
            "Artificial Intelligence enables machines to mimic human intelligence.",
            "AI helps systems learn, reason and make decisions."
        ]
    },

    "ml": {
        "patterns": ["machine learning", "ml"],
        "responses": [
            "Machine Learning is a subset of AI that learns patterns from data.",
            "ML allows systems to improve automatically through experience."
        ]
    },

    "deep_learning": {
        "patterns": ["deep learning", "cnn", "ann", "lstm"],
        "responses": [
            "Deep Learning uses neural networks with multiple hidden layers.",
            "CNNs are commonly used for image processing while LSTMs are used for sequence data."
        ]
    },

    "nlp": {
        "patterns": ["nlp", "natural language processing"],
        "responses": [
            "NLP enables computers to understand and process human language.",
            "NLP combines linguistics, AI and machine learning."
        ]
    },

    "llm": {
        "patterns": ["llm", "large language model", "chatgpt"],
        "responses": [
            "LLMs are trained on massive datasets to understand and generate human-like text.",
            "Examples include GPT, Gemini and Llama."
        ]
    },

    "bye": {
        "patterns": ["bye", "goodbye", "see you"],
        "responses": [
            "Goodbye! Happy learning.",
            "See you soon. Keep building projects!"
        ]
    }
}

# ===========================
# PREPROCESS USER INPUT
# ===========================

def preprocess_text(text):

    text = text.lower()

    # Remove non-ASCII characters using encode/decode to avoid SyntaxError with regex null bytes
    text = text.encode('ascii', 'ignore').decode('ascii')

    tokens = word_tokenize(text)

    stop_words = set(stopwords.words('english'))

    tokens = [word for word in tokens
              if word not in stop_words]

    return tokens

# ===========================
# FIND INTENT
# ===========================

def find_intent(tokens):

    user_text = " ".join(tokens)

    for intent, data in knowledge_base.items():

        for pattern in data["patterns"]:

            if pattern in user_text:
                return intent

    return None

# ===========================
# CHAT HISTORY
# ===========================

chat_history = []

# ===========================
# CHATBOT RESPONSE
# ===========================

def chatbot_response(user_input):

    tokens = preprocess_text(user_input)

    intent = find_intent(tokens)

    if intent:

        response = random.choice(
            knowledge_base[intent]["responses"]
        )

    else:

        response = (
            "Sorry, I couldn't understand. "
            "Ask me about AI, ML, Deep Learning, NLP or LLMs."
        )

    chat_history.append({
        "user": user_input,
        "bot": response
    })

    return response

# ===========================
# SAVE CHAT HISTORY
# ===========================

def save_chat():

    with open(
        "chat_history.json",
        "w",
        encoding="utf-8"
    ) as file:

        json.dump(
            chat_history,
            file,
            indent=4,
            ensure_ascii=False
        )

# ===========================
# MAIN CHAT LOOP
# ===========================

print("=" * 50)
print("Enisteen - NLP Study Buddy")
print("Type 'quit' to exit")
print("=" * 50)

while True:

    user_input = input("\nYou: ")

    if user_input.lower() == "quit":

        save_chat()

        print("\nEnisteen: Goodbye! Chat history saved.")

        break

    response = chatbot_response(user_input)

    print("\nEnisteen:", response)
```
